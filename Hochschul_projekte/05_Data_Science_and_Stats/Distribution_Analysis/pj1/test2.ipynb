```python
from scipy.integrate import quad

# Beispiel-Dichtefunktion: Gleichverteilung von 0 bis 240
def f(x):
    if 0 <= x <= 240:
        return 1 / 240
    else:
        return 0

# Intervallgrenzen, z. B. 162 bis 200
a = 162
b = 200

# Fläche unter der Kurve berechnen = Wahrscheinlichkeit
wahrscheinlichkeit, _ = quad(f, a, b)

print(f"Die Wahrscheinlichkeit, dass X zwischen {a} und {b} liegt, beträgt ca. {wahrscheinlichkeit:.3f}")
```
