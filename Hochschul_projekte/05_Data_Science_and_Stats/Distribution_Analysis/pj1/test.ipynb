```python
import matplotlib.pyplot as plt
from collections import Counter
import itertools

# Alle möglichen Ergebnisse (Kopf = 1, Zahl = 0)
würfe = list(itertools.product([0, 1], repeat=3))  # 3 Würfe

# Anzahl der Köpfe zählen
anzahl_kopf = [sum(wurf) for wurf in würfe]

# Häufigkeiten zählen
häufigkeit = Counter(anzahl_kopf)

# Wahrscheinlichkeiten berechnen
gesamt = len(anzahl_kopf)
wahrscheinlichkeit = {k: v / gesamt for k, v in häufigkeit.items()}

# Darstellung
plt.bar(wahrscheinlichkeit.keys(), wahrscheinlichkeit.values(), color='skyblue')
plt.title("Wahrscheinlichkeiten für Anzahl Kopf bei 3 Würfen")
plt.xlabel("Anzahl Kopf")
plt.ylabel("Wahrscheinlichkeit")
plt.xticks([0, 1, 2, 3])
plt.grid(axis='y')
plt.show()
```
